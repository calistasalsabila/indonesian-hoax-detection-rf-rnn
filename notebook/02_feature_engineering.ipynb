## 2. Feature Engineering

### 2.1 Importing the Library

```python
import pickle
import numpy as np 
from sklearn.feature_extraction.text import TfidfVectorizer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from imblearn.over_sampling import SMOTE
```

### 2.2 Load the Dataset

```python
X_train = pickle.load(open("../dataset/processed/01_after_preprocessing/X_train.pkl", "rb"))
X_test  = pickle.load(open("../dataset/processed/01_after_preprocessing/X_test.pkl", "rb"))
y_train = pickle.load(open("../dataset/processed/01_after_preprocessing/y_train.pkl", "rb"))
y_test  = pickle.load(open("../dataset/processed/01_after_preprocessing/y_test.pkl", "rb"))
```

### 2.3 Define the Parameter

```python
VOCAB_SIZE = 5000
TFIDF_FEATURES = 5000
MAX_LEN = 200
```

### 2.4 TF-IDF

```python
tfidf = TfidfVectorizer(max_features=TFIDF_FEATURES)
tfidf.fit(X_train)

X_train_tfidf = tfidf.transform(X_train).toarray() 
X_test_tfidf  = tfidf.transform(X_test).toarray()
```

### 2.5 Tokenizer and Padding

```python
tokenizer = Tokenizer(num_words=VOCAB_SIZE)
tokenizer.fit_on_texts(X_train)

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq  = tokenizer.texts_to_sequences(X_test)

X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LEN)
X_test_pad  = pad_sequences(X_test_seq,  maxlen=MAX_LEN)
```

### 2.5 Label Conversion

```python
y_train = np.array(y_train, dtype="float32")
y_test  = np.array(y_test, dtype="float32") 

print(f"Shape X_train_pad: {X_train_pad.shape}")
print(f"Shape X_train_tfidf: {X_train_tfidf.shape}")
print(f"Shape y_train: {y_train.shape}")
```

### 2.6 Class Balancing

```python
y_train_np = np.array(y_train, dtype="float32") 
y_test_np  = np.array(y_test, dtype="float32")
```

```python
print("Data training before SMOTE")
print("Bentuk X_train_tfidf:", X_train_tfidf.shape)
print("Bentuk y_train_np:", y_train_np.shape)

smote = SMOTE(random_state=42)

X_train_resampled, y_train_resampled = smote.fit_resample(X_train_tfidf, y_train_np)

print("\nData training after SMOTE")
print("Bentuk X_train resampled:", X_train_resampled.shape)
print("Bentuk y_train resampled:", y_train_resampled.shape)
```

### 2.6 Save the Dataset

```python
pickle.dump(tfidf, open("../dataset/processed/02_after_FE/tfidf_vectorizer.pkl", "wb"))
pickle.dump(tokenizer, open("../dataset/processed/02_after_FE/tokenizer.pkl", "wb"))

pickle.dump(X_train_resampled, open("../dataset/processed/02_after_FE/X_train_tfidf.pkl", "wb")) 
pickle.dump(X_test_tfidf,  open("../dataset/processed/02_after_FE/X_test_tfidf.pkl", "wb")) 
pickle.dump(X_train_pad, open("../dataset/processed/02_after_FE/X_train_pad.pkl", "wb")) 
pickle.dump(X_test_pad,  open("../dataset/processed/02_after_FE/X_test_pad.pkl", "wb")) 
pickle.dump(y_train_resampled, open("../dataset/processed/02_after_FE/y_train.pkl", "wb")) 
pickle.dump(y_test_np,  open("../dataset/processed/02_after_FE/y_test.pkl", "wb")) 
```
